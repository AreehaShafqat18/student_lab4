# Part 6: Indexing and Slicing


```python
import numpy as np

scores = np.array([
    [45, 78, 90, 60],
    [32, 65, 70, 55],
    [88, 92, 79, 95]
])

print(scores[0, 1])       # Row 0, column 1
print(scores[-1, -2])     # Last row, second-last column
print(scores[:, 0])       # First column of every row
print(scores[0:2, 1:3])   # Rows 0-1, columns 1-2

# A slice is a view
view = scores[0, :2]
view[0] = 0
print("Original after editing view:", scores[0])

# Restore the original value
scores[0, 0] = 45

# A copy does not change the original
copy = scores[1, :2].copy()
copy[0] = 999
print("Original after editing copy:", scores[1])
```

# Part 7: Reshaping Arrays

```python
import numpy as np

a = np.arange(12)

print(a.reshape(3, 4))
print(a.reshape(2, -1).shape)  # NumPy calculates the missing dimension
print(np.array([1, 2, 3]).reshape(3, 1))

try:
    a.reshape(5, 3)
except ValueError as e:
    print("Error:", e)
```

# Part 8: Broadcasting

```python
import numpy as np

# One bonus per subject
bonus = np.array([5, 0, 10, 0])
print(scores + bonus)

# Broadcasting a column and a row
col = np.array([[1], [2], [3]])
print(col + np.array([10, 20, 30]))

# Subtract each subject's mean
means = scores.mean(axis=0)
print(np.round(scores - means, 1))
```

## Task 1: Test Broadcasting Shapes

```python
import numpy as np

shape_pairs = [
    ((4, 3), (3,)),
    ((4, 3), (4,)),
    ((4, 1), (1, 5))
]

for shape1, shape2 in shape_pairs:
    try:
        result = np.ones(shape1) + np.ones(shape2)
        print(f"{shape1} + {shape2}: Works, result shape = {result.shape}")
    except ValueError as e:
        print(f"{shape1} + {shape2}: Error - {e}")
```

# Part 9: Vectorization and Aggregation

```python
import numpy as np

marks_list = [40, 60, 72]

# Using a loop
with_loop = []
for m in marks_list:
    with_loop.append(m / 80 * 100)

# Using vectorization (without a loop)
vectorized = np.array(marks_list) / 80 * 100

print(with_loop, vectorized)

# Basic aggregation
print(scores.sum(), scores.mean(), scores.min(), scores.max())

# Sum of each subject
print(scores.sum(axis=0))

# Sum of each student
print(scores.sum(axis=1))

# Standard deviation
print(np.round(scores.std(), 2))
print(np.round(scores.std(ddof=1), 2))

# Compare every score with 60
print(scores >= 60)
```
