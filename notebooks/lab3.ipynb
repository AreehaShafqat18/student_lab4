# Part 5: Arrays, Dimensions and Shapes

```python
import numpy as np

scores = np.array([
    [45, 78, 90, 60],
    [32, 65, 70, 55],
    [88, 92, 79, 95]
])

print(scores.ndim, scores.shape, scores.size, scores.dtype)
print(np.zeros((2, 3)))
print(np.arange(0, 10, 2), np.linspace(0, 1, 5))

lst = [1, 2, 3]
arr = np.array(lst)

print(lst * 2, arr * 2)
```

## Task 1: Create NumPy Arrays

```python
import numpy as np

ones_array = np.ones((2, 3))
full_array = np.full((2, 2), 7)
even_numbers = np.arange(2, 11, 2)

print("Ones array:")
print(ones_array)

print("Full array:")
print(full_array)

print("Even numbers:")
print(even_numbers)
```

## Task 2: Dimensions, Shape and Size

```python
import numpy as np

arr = np.zeros((2, 3, 4))

# Prediction:
# ndim = 3
# shape = (2, 3, 4)
# size = 24

print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
```
