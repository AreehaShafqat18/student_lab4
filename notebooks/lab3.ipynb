# Part 6: Indexing and Slicing


```python
import numpy as np

scores = np.array([
    [45, 78, 90, 60],
    [32, 65, 70, 55],
    [88, 92, 79, 95]
])

print(scores[0, 1])       # Row 0, column 1
print(scores[-1, -2])     # Last row, second-last column
print(scores[:, 0])       # First column of every row
print(scores[0:2, 1:3])   # Rows 0-1, columns 1-2

# A slice is a view
view = scores[0, :2]
view[0] = 0
print("Original after editing view:", scores[0])

# Restore the original value
scores[0, 0] = 45

# A copy does not change the original
copy = scores[1, :2].copy()
copy[0] = 999
print("Original after editing copy:", scores[1])
```
